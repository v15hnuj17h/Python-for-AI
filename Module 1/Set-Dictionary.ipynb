```python
#Define sets

set1 = {1, 2, 3}
set2 = {3, 2, 1}
set3 = {1, 2}
set4 = {3, 4}
set5 = {2, 3, 4}

#Equality and inequality

print(set1 == set2)
print(set1 != set2)
```

```python
#Subset and superset

print(set1 < set2)
print(set1 > set3)
```

```python
#Disjoint set

print(set1.isdisjoint(set4))
print(set3.isdisjoint(set4))
```

```python
intersection = set1 & set5
print(intersection)
print(intersection <= set1)

union = set2 | set4
print(union)
print(union >= set2)
```

```python
# Define sets of students
math_students = {"Alice", "Bob", "Charlie", "David"}
science_students = {"Charlie", "David", "Eve", "Frank"}

# Display the sets
print(f"Math Students: {math_students}")
print(f"Science Students: {science_students}")

# Union: Students who are in either Math or Science
all_students = math_students | science_students
print(f"Students in either Math or Science: {all_students}")

# Intersection: Students who are in both Math and Science
both_classes = math_students & science_students
print(f"Students in both Math and Science: {both_classes}")

# Difference: Students who are in Math but not in Science
only_math = math_students - science_students
print(f"Students in Math but not in Science: {only_math}")

# Symmetric Difference: Students who are in either Math or Science, but not both
exclusive_students = math_students ^ science_students
print(f"Students in either Math or Science, but not both:{exclusive_students}")

# Check if all Math students are also in Science
is_subset = math_students <= science_students
print(f"Are all Math students also Science students? {is_subset}")

# Add a new student to Math
math_students.add("Grace")
print(f"Updated Math Students: {math_students}")

# Remove a student from Science
science_students.discard("Frank")
print(f"Updated Science Students: {science_students}")
```

```python
# Using pop() - Removes the specified key and returns its value

dict1 = {"a": 1, "b": 2, "c": 3}

value = dict1.pop("b")

print(value)
print(dict1)


# Using del - Deletes a key-value pair

del dict1["a"]
print(dict1)  # Output: {'c': 3}
```

```python
# Using popitem() - Removes the last inserted key-value pair

dict1 = {"x":10, "y":20}
last_item = dict1.popitem()
print(last_item)
print(dict1)


# Using clear() - Removes all items
dict1.clear()
print(dict1)  # Output: {}
```

```python
# Telephone directory program

#Add a contact 
#Update a contact
#Delete a contact
#Search for a contact
#Display all contact
#Exit the program


directory = {}

while True:
    print("\nTelephone Directory Menu:")
    print("1. Add a Contact")
    print("2. Update a Contact")
    print("3. Delete a Contact")
    print("4. Search for a Contact")
    print("5. Display All Contacts")
    print("6. Exit\n")
    
    choice = input("Enter your choice: ")
    if choice == '1':
        contact = input("Enter the contact name: ")
        if contact in directory:
            print("Contact already exists")
        else:
            directory[contact] = int(input("Enter the number: "))
            print("Contact added successfully")
    elif choice == '2':
        contact = input("Enter the contact name: ")
        if contact not in directory:
            print("Contact does not exists")
        else:
            directory[contact] = int(input("Enter the new number: "))
            print("Number updated successfully")
    elif choice == '3':
        contact = input("Enter the contact name: ")
        if contact not in directory:
            print("Contact does not exists")
        else:
            del directory[contact]
            print("Contact deleted successfully")
    elif choice == '4':
        contact = input("Enter the contact name: ")
        if contact not in directory:
            print("Contact does not exists")
        else:
            print(f"{contact} : {directory[contact]}")
    elif choice == '5':
        print(directory)
    elif choice == '6':
        print("Exiting the program")
        break
    else:
        print("Input must be in 1 to 6")
```

```python
# This program allows you to manage the stock of products in a store using a dictionary. The 
# product name will serve as the key, and its stock quantity will serve as the value. The program 
# provides the following functionalities:
#  1. Add a New Product: Add a new product to the inventory.
#  2. Update the Stock: Update the stock quantity of an existing product.
#  3. Check the Stock: View the stock quantity of a specific product.
#  4. Display All Products: List all products and their stock levels.
#  5. Exit the Program: Exit the program


inventory = {}

while True:
    print("\n\nStore Inventory Menu:")
    print("1. Add a New Product")
    print("2. Update the Stock of a Product")
    print("3. Check the Stock of a Product")
    print("4. Display All Products")
    print("5. Exit\n")
    choice = input("Enter your choice (1-5): ")
    if choice == '1':  # Add a New Product
        product = input("Enter the product name: ")
        if product in inventory:
            print(f"The product '{product}' already exists in the inventory.")
        else:
            stock = int(input("Enter the initial stock quantity: "))
            inventory[product] = stock
            print(f"Product '{product}' added with a stock of {stock}.")
    elif choice == '2':  # Update the Stock
        product = input("Enter the product name to update: ")
        if product in inventory:
            stock = int(input(f"Enter the new stock quantity for '{product}': "))
            inventory[product] = stock
            print(f"Stock of '{product}' updated to {stock}.")
        else:
            print(f"Product '{product}' not found in the inventory.")
    
    elif choice == '3':  # Check the Stock
        product = input("Enter the product name to check: ")
        if product in inventory:
            print(f"The stock of '{product}' is {inventory[product]}.")
        else:
            print(f"Product '{product}' not found in the inventory.")
    elif choice == '4':  # Display All Products
        if inventory:
            print("\nCurrent Inventory:")
            for product, stock in inventory.items():
                print(f"{product}: {stock}")
        else:
            print("The inventory is empty.")
    elif choice == '5':  # Exit
        print("Exiting the program. Goodbye!")
        break
    else:
        print("Invalid choice")
```
